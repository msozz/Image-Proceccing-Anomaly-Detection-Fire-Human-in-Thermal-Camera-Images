```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Görseli yükle (örnek: termal kamera görüntüsü) ---
img = cv2.imread('thermal_fire.jpg')

# Görüntü çok büyükse küçült (isteğe bağlı ama önerilir)
img = cv2.resize(img, (640, 480))

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# --- 2. Histogram Equalization (kontrast artırma) ---
equalized = cv2.equalizeHist(gray)

# --- 3. Saliency Map (makaledeki yöntem) ---
saliency = cv2.saliency.StaticSaliencySpectralResidual_create()
(success, saliency_map) = saliency.computeSaliency(equalized)
saliency_map = (saliency_map * 255).astype("uint8")

# --- 4. Sobel Filter (kenar tespiti) ---
sobelx = cv2.Sobel(equalized, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(equalized, cv2.CV_64F, 0, 1, ksize=3)
sobel_edges = cv2.magnitude(sobelx, sobely)
sobel_edges = np.uint8(sobel_edges / np.max(sobel_edges) * 255)

# --- 5. Sonuçları göster ---
plt.figure(figsize=(15,4))

plt.subplot(1, 4, 1)
plt.title("Original (Thermal)")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title("Histogram Equalized")
plt.imshow(equalized, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title("Saliency Map")
plt.imshow(saliency_map, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title("Sobel Edges")
plt.imshow(sobel_edges, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
